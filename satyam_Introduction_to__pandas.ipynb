**DataFrame.sample(n=None, frac=None, replace=False, weights=None, random_state=None, axis=None, ignore_index=False)**
these are following attribute of sample function
1. n👉 Number of rows to select
2. frac👉 Fraction (percentage) of data
3. replace👉 Allow repetition or not
False → no repeat (default)
True → repeat allowed
4. weights👉 Probability for each row
5. random_state👉 Fix randomness (for same result every time)
6. axis👉 Sample rows or columns
0 → rows (default)
1 → columns
ignore_index
👉 Reset index or not

## Groupby
* ## group have three operation
* **Split** - we split datafram into smaller dataframe based on the value of keys
* **Apply** - we apply desired
aggregation/trandsformation on each dataframe
* **Combine**- we combine result from apply state into a datafram



```python
df = pd.DataFrame({'key':['A','B','C']*2, # string ABCABC
                   'data': range(6)})
```

```python
df
```

```python
df.groupby('key').sum()
```

```python
df.groupby(by='key').sum()
```

```python
df.groupby(level='data').sum()
```
